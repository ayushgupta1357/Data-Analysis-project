```python
folder="C:/Users/wwway/Desktop/Data Science/DATA ANALYTICS PROJECT 1/data/"
```

```python
#engine initialization
engine=create_engine("mysql+pymysql://root:AYUSHgupta12345@localhost:3306/")

with engine.connect() as conn:
    conn.execute(text('create database if not exists Sales_DataBase'))  #database creation if not exist
```

```python
sales_engine=create_engine("mysql+pymysql://root:AYUSHgupta12345@localhost:3306/Sales_Database")   #Engine to connect  to the database
```

```python
#ingestion function to ingest files to sql database
def ingest_db(df,tablename,eng):
    df.to_sql(tablename,con=eng,if_exists="replace",index=False)
```

```python
#Reading and ingesting files
start_time=time.time()
for file in os.listdir(folder):
    if file.endswith('.csv'):
        try:
            df=pd.read_csv(folder+file)
            ingest_db(df,file[:-4],sales_engine)
            logging.info(file+" Ingested")
            print(f"FIle Inserted {file}")
        except Exception as e:
            print(e,'Trying with python engine..........')
            
            try:
                df=pd.read_csv(folder+file,engine="python",on_bad_lines='skip')
                ingest_db(df,file[:-4],sales_engine)
                logging.info(f"File Inserted {file}")
                print(file," Inserted")
            except:
                print(file,' is corrupted!!!!')

endtime=time.time()       
timetaken=(endtime-start_time)/60 
logging.info(f"Total time taken for ingestion is :{timetaken}")
print("Sucessfully Ingested ")
        
```
